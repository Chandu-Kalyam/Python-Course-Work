# Feature Engineering

```python
# Def: Feature engineering helps to improve the performance of ML models.
# ML models works as mathematical equations.
# Mathematical equations cannot be apply on categorical data.
# In FE we convert categorical data to numerical data.

# Feature scaling: to bring the data in a particular range.
```

### Feature Encoding

```python
# converting categorical data to numerical data.
# Popular method for feature encoding:
# 1. Label encoding
# 2. Ordinal encoding
# 3. One hot encoding
```

# 1. Label encoding

```python
# def: Assign an integer to each category in lexographically order.
# Example-1:
from sklearn.preprocessing import LabelEncoder
import pandas as pd

data = {"color" : ["white", "red", "yellow", "blue", "green"]}
df = pd.DataFrame(data)
df
# create the LabelEncoder object
encoder = LabelEncoder()

# fit() function allot the integer value to each category
encoder.fit(df["color"])
# transform() function transform the actual data based upon fit() function allocation.
df["encoded_color"] = encoder.transform(df["color"])
print(df)
```

# 2. Ordinal encoding

```python
# convert categorical data into numerical data based upon the ranking.
from sklearn.preprocessing import OrdinalEncoder

data = [["Phd"], ["UG"], ["MS"], ["UG"], ["Phd"]]
# categories we can provides as ranking categories = [["UG", "MS", "Phd"]] so now ug as 0, ms as 1 and phd as 2
encoder = OrdinalEncoder(categories = [["UG", "MS", "Phd"]])

encoder_data = encoder.fit_transform(data)
print(encoder_data)
```

```python
# Use label encoder when we don't have ranking in the categories.
# Use Ordinal encoder when we have ranking in the categories.
```

# One-hot encoder

```python
# convert categorical data into binary indicators.

data = {"country" : ["India", "USA", "Uganda", "Japan", "Thailand", "Nigeria", "India"]}
df = pd.DataFrame(data)

encoded_data = pd.get_dummies(df,dtype=int)
encoded_data
```

```python
df
```

```python
# Machine love  the one hot encoder due to binary encoding.
# Don't use one hot encoder, when number of categories are large
# Use one hot encoder when having less number of categories
```

# Method-2: one hot encoding using sklearn

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output = False,dtype=int)
encoded_data = encoder.fit_transform(df[["country"]])
print(encoded_data)
```
