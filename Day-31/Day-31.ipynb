# Data filtering and indexing

```python
# Query: find those records where salary > 10000

df[df["salary"] > 100000]
```

```python
df[df["salary"] < 50000]
```

```python

df[(df["salary"] > 50000) & (df["salary"] < 100000)]
```

```python

df[(df["salary"] > 50000) & (df["salary"] < 100000)].shape
```

```python
# Query: find those records wheere salary < 50000 or salary >150000
df[(df["salary"] < 50000) | (df["salary"] > 150000)]
```

```python
# Query: find the record of the person who get maximum salary.
df[df["salary"] == df["salary"].max()]
```

```python
# find the record of the person who got max marriage proposals
df[df["Marriage_proposal_count"] == df["Marriage_proposal_count"].max()]
```

```python
# Query: find the record of the person who got minimum salary
df[df["salary"] == df["salary"].min()][['Name', "Profession"]]
```

# Indexing

```python
# loc and iloc
```

```python
# Query: Print 1 to 6 records and Name to profession columns
df.columns
```

```python
# Query: Print 1 to 6 records and Name to profession columns
# df.loc[row: columns]
df.loc[1:6, "Name" : "Profession"]
```

```python
# Query: Print 10 to 20 records and Age to salary columns
df.loc[10:20:2, ["Age","salary"]]
```

```python
df.iloc[1:6, 0 : 5]
```

```python
# iloc
# Query: print record from 1 to 6 and columns from 1 to 3
df.iloc[0:7, 1:4]
```

```python
# Interview question: What is the difference loc and iloc
# in loc we need to mention the column names
# while in iloc we mention the index of columns
# iloc goes up to n-1, but loc goes upto nth value

# iloc is little faster than loc
# because loc need to convert column names to index
```

```python
df.iloc[0,1]
```

```python
df.iloc[0,2]
```

```python
df.at[0, "Name"]
```

```python
df.iat[0,1]
```

```python
# at & iat used specifically to access single element
# at & iat are faster than loc & iloc
```
