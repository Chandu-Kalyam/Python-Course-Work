# 3. Assignment Operator (+=, -=, *=, /=,....)

```python
a = 10
a +=5

print(a)
a -=5

print(a)
a *=5

print(a)
a /=5
print(a)

a %=5
print(a)
```

```python
a = 10
b = 3
a &= b
print(a)
```

```python
a |=b
print(a)
```

```python
a <<=b
print(a)
```

```python
a >>=b
print(a)
```

```python
#  walrus operator  (:=)  
# walrus operator will first assign the value to the variable then pass it to outer function
print(x:=2)
# print(x=2)  typeerror 
```

4. Comparision operator (==, !=, <=, >=,<,>)

```python
a = 10
b = 5
print(a == b)
print(a != b)
```

# Logical operator

```python
# and , or, not
a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

# Identity operator (is, is not)

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(a == b)
print(a is b)
print(a is c)
print(a is not c)
```

# Membership operators

```python
#  in and not in
fruits = ['Apple', "Mango", "Watermeleon", "Jackfruit"]
print("Apple" in fruits)
print("kiwi" not in fruits)
print("Apple" not in fruits)
```
