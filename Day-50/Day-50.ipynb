```python
# SVM is a supervised ml algorithm.
# SVM is mainly used for classification based problems.
# Although we can use it for rergression based problem also.
# SVM can be used for Binary/Multiclass classification.
# SVM takes time to execute. Its not good idea to use it for large datasets.
```

```python
# libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC  # SVC: support vector classifer
from sklearn.datasets import load_iris
```

```python
# We need to find type of iris flower
# based upon sapal length, sapalwidth and petal length and petal width
```

```python
# loading the dataset
iris = load_iris()
```

```python
# Describe the dataset
dir(iris)
```

```python
iris["feature_names"]
```

```python
print(iris.DESCR)
```

```python
# store the iris flower dataset into dataframe format
x  = pd.DataFrame(iris.data, columns = iris["feature_names"])
x
```

```python
y = iris.target
```

```python
# split the data into training and testing data
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.25, random_state = 42)
```

```python
print(len(x_train))
```

```python
# train the model
model = SVC()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict([[4.9,3.2, 1.3,0.2]])
```
