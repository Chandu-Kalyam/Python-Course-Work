# Enumerate function

```python
# enumerate() function add the counter (index) with the list items.
```

```python
fruits = ['Mango', 'Apple', 'Jackfruit', 'Guava']
for index, fruit in enumerate(fruits):
    print(f"{index} : {fruit}")
```

```python
for index, fruit in enumerate(fruits, start = 1):
    print(f"{index} : {fruit}")
```

# Zip() function

```python
# zip function is used to combine/zip multiple list/tuple items, element wise.
```

```python
Hero = ['Mahesh', 'Mallikarjun', 'Pavan Kalyan','Balaji']
Heroine = ['Anushka', 'Samanta', 'Rashmika', 'Kajal']
Movie = ['bahubali','rrr','athadu']
pairs = zip(Hero, Heroine, Movie)
print(list(pairs))
```

```python
fruits = ['apple', 'mango']
vege = ['carrot','mirchi']
zipper = zip(fruits, vege)
for fruit, veg in zipper:
    print(fruit, veg)
```

# Decorators

```python
def decorator_fun(func):         # func => cake()
    def wrapper_fun():          # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
        print("Add cherry")      # Extra functionality
        print("Add choco chips")
    return wrapper_fun

def cake():
    print("Take the flour")
    print("Add eggs and other ingredients")
    print("Cook the cake in oven")
    

decorated_function = decorator_fun(cake)
decorated_function()
```

```python
# Example-2 use of @ to represent the decorator
def decorator_fun(func):         # func => cake()
    def wrapper_fun():          # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
        print("Add cherry")      # Extra functionality
        print("Add choco chips")
    return wrapper_fun

@decorator_fun
def cake():
    print("Take the flour")
    print("Add eggs and other ingredients")
    print("Cook the cake in oven")
    
cake()
```

```python
# Multiple decorator for some function

def decorator_fun1(func):         # func => cake()
    def wrapper_fun():          # wrapper function add the extra functionality
        print("Buy the ingredients")
        func()
    return wrapper_fun

def decorator_fun2(func):         # func => cake()
    def wrapper_fun():          # wrapper function add the extra functionality
        func()
        print("Add cherry")      # Extra functionality
        print("Add choco chips")
    return wrapper_fun

@decorator_fun1
@decorator_fun2
def cake():
    print("Take the flour")
    print("Add eggs and other ingredients")
    print("Cook the cake in oven")
    
cake()
```
