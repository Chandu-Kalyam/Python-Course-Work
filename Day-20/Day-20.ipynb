# 1. Single Inheritance: One parent One child

```python
# Example-1: Structure of Single Inheritance
# Parent class
class Hyd_food:
    TASTE = "Very very very Spicy"
    
    def biryani(self):
        print("Very very tasty biryani")
    
# Child class
class PistaHouse(Hyd_food):
    pass

p = PistaHouse()      # It is object of child class
# child class object can access the properties & function of parent class

print(p.TASTE)
print()
p.biryani()
```

```python
# Example-2: super() function
# super() : used to access the parent properties and function from inside the child class
# Parent class
class Hyd_food:
    TASTE = "Very very very Spicy"
    
    def biryani(self):
        print("Very very tasty biryani")
    
# Child class
class PistaHouse(Hyd_food):
    def display(self):
        super().biryani()
        print(super().TASTE)

p = PistaHouse()      # It is object of child class
p.display()
```

# 2. Multiple Inheritance: Many parents one child

```python
# Example-1: Design structure
class A:
    def __init__(self):
        print("I am A class")

    def display(self):
        print("I am a class")

class B:
    def __init__(self):
        print("I am B class")
    
    def show(self):
        print("I am b class")
    
class C(A,B):
    def __init__(self):
        print("I am C class")
        A.__init__(self)
        B.__init__(self)
        
        
c = C()
print()
c.display()
c.show()
print(isinstance(c,A) , isinstance(c,B))
```

```python
# Example-2: Design the structure
# Sequence in which parent are assign that matter.
class A:
    def __init__(self):
        print("I am A class")

    def display(self):
        print("I am a class")

class B:
    def __init__(self):
        print("I am B class")
    
    def display(self):
        print("I am b class")
    
class C(A,B):
    def __init__(self):
        print("I am C class")
        A.__init__(self)
        B.__init__(self)
        
        
c = C()
print()
c.display()
print(isinstance(c,A) , isinstance(c,B))
```

```python
# Example-3: constructor function
# Sequence in which parent are assign that matter.
class A:
    def __init__(self):
        print("I am A class constructor")

class B:
    def __init__(self):
        print("I am B class constructor")
    
    
    
class C(A,B):  # first priority of parent class is A
    def __init__(self):
        print("I am C class constructor")
        A.__init__(self)
        B.__init__(self)
        
        
c = C()
print()
print(isinstance(c,A) , isinstance(c,B))
```

```python
# Example-4.1: Design the structure super()
# Sequence in which parent are assign that matter.
class A:
    def __init__(self):
        print("I am A class constructor")

    def display(self):
        print("I am a class")

class B:
    def __init__(self):
        print("I am B class constructor")
    
    def show(self):
        print("I am b class")
    
class C(A,B):
    def __init__(self):
        print("I am C class constructor")
        A.__init__(self)
        B.__init__(self)
        
    def hello(self):
        super().display()
        super().show()
        
        
c = C()
print()
c.display()
print()
c.hello()
print(isinstance(c,A) , isinstance(c,B))
```

```python
# Example-4.1: Design the structure super()
# super() follow the MRO approach (Method Resolution Order)
# Sequence in which parent are assign that matter.
class A:
    def __init__(self):
        print("I am A class constructor")

    def display(self):
        print("I am a class")
        super().display()

class B:
    def __init__(self):
        print("I am B class constructor")
    
    def display(self):
        print("I am b class")
    
class C(A,B):
    def __init__(self):
        print("I am C class constructor")
        A.__init__(self)
        B.__init__(self)
        
    def hello(self):
        super().display()
        
        
c = C()
print()
c.display()
print()
c.hello()

# super() follow the MRO approach (Method Resolution Order)
print([i.__name__ for i in C.__mro__])
```
