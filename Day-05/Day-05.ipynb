```python
#  Recieving multiple inputs we can use split()
x1, x2 , x3 = input("Enter three names:").split()
print(x1, x2 , x3)
```

```python
x1, x2 , x3 = input("Enter three names:").split(',')
print(x1, x2 , x3)
```

```python
x1, x2 , x3 = map(int, input('Enter three numbers:').split())
print(x1, x2 , x3)
```

```python
x1, x2 , x3 = map(float, input('Enter three numbers:').split())
print(x1, x2 , x3)
```

```python
list = list(map(int, input("Enter any number values").split()))
print(list)
```

# Output in python

```python
# 1. concatenation using commas
# 2. String concatenation using +
# 3. str.format() method
# 4. f-strings method (python 3.6+ version)
# 5. Escape sequences
# 6. end and sep parameters in print()
```

```python
# 1. concatenation using commas
Hero_name = "Prabhas"
age = 40
fee = 100
print(Hero_name,age,fee)
```

```python
# 2. String concatenation using +
Hero_name = "Prabhas"
Heroine_name = 'Anushka'
fee = 100
print(Hero_name + " " + Heroine_name+" " + str(fee))
```

```python
# 3. str.format() method
Hero_name = "NTR"
age = 40
print("Name:{1} Age:{0}".format(Hero_name,age))
```

```python
# 4. f-strings method (python 3.6+ version)
print(f"Name: {Hero_name}, Age: {age}")
```

```python
# 5. Escape sequences in print() \n \n \t
Hero_name = "Prabhas"
Heroine_name = 'Anushka'
fee = 100
print(f"Hero: \"{Hero_name}\" \nHeroine_name: \"{Heroine_name}\" \nFee: {fee}")
```

```python
# 6. end and sep parameters in print()

print(Hero_name, end=" ")
print(Heroine_name)
```

```python
print(Hero_name, Heroine_name, fee, sep="#")   # by default value of sep = space " "
arr = ['prabhas', 'ntr', 'alluarjun']
print(*arr, sep="\n")
```

```python
# end parameter is used to decide how the print statement will END.
# sep parameter decides how the multiple values will be separated.
```
