```python
#  bitwise opereator works on binary numbers
#  bitwise operator are faster than arthimatic operator
#  bitwise operator perform many operations in easy way due to binary numbers.
```

```python
print(50 * 64)
print(50<< 6)       # 
```

```python
a = 10
b = 3       
                                    # 1010 and 0011  -->  0010  ==> 2         10 and 10  --> 10  ==> 2   
print(a & b)                         # 1010 or 0011  -->  1011  ==> 11
print(a | b)                         # 1010 xor 0011  -->  1001  ==> 9
print(a ^ b)
print(a^b)
```

### left shift (<<) , Right shift (>>)

```python
#  Def left shift: Insert new bits(0) to the LSB
# Def right shift: Delete new bits to the LSB
```

```python
#  left shift is multiplication
a = 10                                  # 0000 1010 << 1  ==>  10100  ==> 20    formula: a * 2**k    k is 1  10 * 2**1 ==> 20
print(a<<1)
```

```python
print(a << 2)  
print(a)
```

```python
# Right shift is division
print(a>>1)                             #  0000 1010 >> 1 --> 0000 0101   ==> 5    formula:floor value a/2**k
```

```python
# Bitwise not(~)
a = 12                                   #  a = ~(1100) ==> complement 1100 --> 0011 + 1 ==> 0100  ==> 1011
print(~a)
```
