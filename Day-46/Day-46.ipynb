# Feature scaling: to bring the data in a particular range

```python
# ML models give more importance to large values.
# So we bring all the data into same range. (0 to 1)
# Types of feature scaling:
# 1. min-max scaling
# 2. Standardization (Z-score scaling)
# 3. Robust scaling
```

# 1. Min-Max scaling

```python
# Min-max scaler bring the value by default in 0 to 1 range.
# X_scaled = (X - X_min) / (X_max - X_min)

# X = [100000, 200000,300000,400000, 500000]
# x = 100000 => 0
# x = 200000 => 0.25
# X = 300000 => 300000 - 100000 / (500000 - 100000) = 0.5
# x = 400000 => 0.75
# x = 500000 => 1
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
scaler = MinMaxScaler()
data = [100000, 200000,300000,400000, 500000]
df = pd.DataFrame(data, columns = ["values"])
df
df["scaled_values"] = scaler.fit_transform(df[["values"]])
df

# MinMaxScaler cannot handle the outliers
```

# Standardization (Z-Score scaler)

```python
# z-score scaler is used when data is Normally distributed
# z_score = (x - mean) / SD

# z_score scaler will bring the value in the range of -3 to +3

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data = [100, 200, 300, 400, 500]
df = pd.DataFrame(data, columns = ["values"])
df

df["scaled_values"] = scaler.fit_transform(df[["values"]])
df
```

```python
import numpy as np
mean_value = np.mean(df["values"])
std_value = np.std(df["values"])
x = 100
Z_score = (x - mean_value) / std_value
print(Z_score)
```

# Robust Scaler

```python
# Use the Robust scaler when there is outlier in the dataset
# Robust scaler is based upon IQR and median

# x_scaled = (x - median(x) / IQR(x)
# IQR: Inter Quartile Range = Q3 - Q1

from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

data = [100, 200, 300, 400, 500]
df = pd.DataFrame(data, columns = ["values"])
df["scaled_values"] = scaler.fit_transform(df[["values"]])
df
```

```python
# Manually
# x_scaled = (x - median(x) / IQR(x)
Median = np.median(df.values)
Q1 = np.percentile(df.values, 25)
Q3 = np.percentile(df.values, 75)
IQR = Q3 - Q1
x = 100
x_scaled = (x - Median) / IQR
print(x_scaled)
```
