# Group the data based upon profession

```python
df.groupby("Profession")["Salary"].mean()
```

```python
# Find average salary based upon the profession

df.groupby("Profession")["Salary"].mean()
```

```python
# Group the data based upon multiple column
df.groupby(["Profession", "Age"])["Salary"].mean()
```

```python
# agg(): aggregation function

df.groupby(["Profession", "Age"])["Salary"].agg(["mean", "count"])
```

```python
df[["Age", "Salary"]].agg(["mean", "sum", "min", "max"])
```

```python
# apply different aggregation functions to different columns
```

```python
df.agg({
    "Age": ["min", "mean"],
    "Salary": ["max", "mean"]
})
```

```python
# agg(): groupby()
df.groupby("Profession").agg(
    {
    "Age": ["min", "mean"],
    "Salary": ["max", "mean"]
})
```

```python
# Merging of DataFrame
# merge(): this function can be used to merge the dataframes based upon common column.
# merge(): function work like join in sql: inner, left, right, outer.
```

```python
df1 = pd.DataFrame({
    "ID" : [1,2,3],
    "Dream_company": ["google", "microsoft", "Apple"]
})
df2 = pd.DataFrame({
    "ID": [2,3,4],
    "Dream_package": [50,60,40]
})

merge_df = pd.merge(df1, df2, on = "ID")  # by default it shows inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = "ID", how = "left")  # by default it shows inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = "ID", how = "right")  # by default it shows inner join
merge_df
```

```python
merge_df = pd.merge(df1, df2, on = "ID", how = "outer")  # by default it shows inner join
merge_df
```

# join(): mainly used to work with index

```python
df3 = pd.DataFrame({
    "Hero_Name": ["NTR", "Prabhas", "Mahesh_Babu"]},
    index = [1,2,3]
)

df4 = pd.DataFrame({
    "Heroine_Name": ["C", "A", "R"]},
    index = [2,3,4]
)

joined_df = df3.join(df4, how = "inner")
joined_df
```

# concat()

```python
# def: concat() function is used to stack dataframes vertically or horizontally
```

```python
df5 = pd.DataFrame({"ID": [1,2], "Fav_Movie": ["Paradise", "Bahubali"]})
df6 = pd.DataFrame({
    "ID": [3,4],
    "Fav_Movie": ["SpiderMan", "Dhoom2"]
})
df7 = pd.DataFrame({
    "ID": [5,6],
    "Fav_Movie": ["Pushpa", "Badrinath"]
})
# axis = 0: vertically [adding rows],
# axis = 1: Horizontally [adding column]

concat_df = pd.concat([df5, df6,df7], axis = 0, ignore_index = True)
concat_df
```
