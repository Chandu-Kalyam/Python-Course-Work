```python
# Example-1: how to define abstract class, abstract method

from abc import ABC, abstractmethod

# abstract class is defined by the Classname(ABC): to define abstract method we have to use the @abstractmethod decorator
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self):        #abstract method
        pass
    
    @abstractmethod
    def led_screen(self):
        pass
    
    def insurance(self):     # concrete method
        print("Please take Mahindra company insurance.")
    
    
class Hyd_Mahindra(Head_Mahindra):    # child class
    def air_bag(self):        #abstract method
        print("I will use nylon to design the air bags")
        
    def led_screen(self):
        print("Yes We are implementing Sony led screen currently.")
        
    def normal(self):
        print("Lol you are doing good")

class Noida_Mahindra(Head_Mahindra):
    def air_bag(self):
        print("I will use cotton to design the air bags")
    
    def led_screen(self):
        print("Yes we are currently working on Samsung led screen")
    
hyd = Hyd_Mahindra()
hyd.air_bag()
hyd.led_screen()
hyd.insurance()
print()
nod = Noida_Mahindra()
nod.air_bag()
nod.led_screen()

# head_ = Head_Mahindra()
# head.air_bag()
```
