## Different ways to create a dictionary

```python
empty_dict = {}
empty_dict2 = dict()

print(empty_dict2)
print(type(empty_dict))

int_dict = {
    1: "Mango",
    2: "Banana",
    3: "Watermelon"
}

print(int_dict)

float_dict = {
    1.2: "Mango",
    2.5: "Banana",
    3.7: "Watermelon"
}

print(float_dict)

str_dict = {
    "A": 'Mango',
    "B": "Banana",
    "C": "Watermelon"
}
print(str_dict)

het_dict = {
    10: "Mango",
    20.5: "Banana",
}

print(het_dict)

tuple_dict = {
    (10, 20) : "Mango",
    20.5 : "Banana",
    "C" : "Watermelon",
}
print(tuple_dict)

# list_dict = {
#     [10,20] : "Mango",
#     20 : "Banana"
# }

# Tuple can be used as a key
# list and set cannot be used as key
# Keys must be immutable
# values can be mutable and immutable

dict_comp = {x: x**2 for x in range(6)}
print(dict_comp)
```

```python
int_dict = {1: "Mango", 1: "Mamidi", 2: "Banana"}
print(int_dict)
```

```python
# list of tuple into Dictionary
list_of_tuple = [("Gopichand", 6), ("Chandra", 90), ("Dinesh", 100)]
dict1 = dict(list_of_tuple)
print(dict1)
```

```python
# List of list into Dictionary
list_of_list = [["Gopichand", 6], ["Chandra", 90], ["Dinesh", 100]]
dict1 = dict(list_of_list)
print(dict1)
```

```python
# tuple of tuple into Dictionary
tuple_of_tuple = (("Gopichand", 6), ("Chandra", 90), ("Dinesh", 100))
dict1 = dict(tuple_of_tuple)
print(dict1)
```

```python
# tuple of list into Dictionary
tuple_of_list = (["Gopichand", 6], ["Chandra", 90], ["Dinesh", 100])
dict1 = dict(tuple_of_list)
print(dict1)
```

```python
# set of list into Dictionary not possible
# set of tuple into dictionary is possible
set_of_tuple = {("Gopichand", 6), ("Chandra", 90), ("Dinesh", 100)}
dict1 = dict(set_of_tuple)
print(dict1)
```

# Accessing dictionary items

```python
drink_items = {
    "thumpsup" : 50,
    "Sugarcane" : 20,
    "Coconut water" : 20,
    "Water" : 20,
    "rosemilk" : 30,
}

val = drink_items['Sugarcane']
print(val)

val2 = drink_items.get("Water")
print(val2)

# val3 = drink_items["softdrink"] this error no key is present
val3 = drink_items.get("softdrink") # no error but outputs None
print(val3)

val4 = drink_items.get("minute maid", "Not available")
print(val4)
```

```python
for key, value in drink_items.items():
    print(key, value)
```

```python
print(drink_items.keys())
```

```python
print(drink_items.values())
```

```python
print(drink_items.items())
```

```python
nested_dict = {
    100 : {
        "Hello" : 10,
        "World" : 20,
    },
    200 : {
        'A' : 50,
        "B" : 60
    },
}

print(nested_dict[100]['World'])
```

# insert and update operations on dictionary


```python
x_dict = {
    'name' : 'Chinki',
    'age' : 80
}

x_dict['age'] = 100


x_dict['city'] = 'Kurnool'     # insert
print(x_dict)

# insert and update using the update() function
# update() can insert and update multiple items together
x_dict.update({'age': 20, 'occupation' : 'IAS', 'name': 'minki', "Salary": 1.5})
print(x_dict)
```

```python
# pop() removes the item whose key is specified
x_dict.pop("Salary")
print(x_dict)
```

```python
x_dict.pop('age')
print(x_dict)
```

```python
# popitem() will delete the last item
x_dict.popitem()
print(x_dict)
```

```python
x_dict.clear()
del x_dict
```
