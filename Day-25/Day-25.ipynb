```python
# open("filename", "mode")
# f is file object throught which we can access the content of file.
f = open("file1.txt", "r")
for line in f:
    print(line)
    
f.close()
```

```python
# with statement thiswill automatically close the file
with open("file1.txt", 'r') as f:
    for line in f:
        print(line)
```

```python
# read(): to get all lines from the file
with open("file1.txt", "r") as f:
    print(f.read())
```

```python
# readlines() : to get all the lines into a list,, to access particular line data.
with open("file1.txt", 'r') as f:
    lines = f.readlines()
    print(lines)
```

```python
# readlines() : to get all the lines into a list,, to access particular line data.
with open("file1.txt", 'r') as f:
    lines = f.readlines()
    print(lines[1])
```

```python
with open("file1.txt", 'r') as f:
    lines = f.readlines()
    print(*lines)  # unpacking of list.
```

# write() function

```python
# To write the content inside the file.
# we will use the "w" mode with write() function.
# if file don't exist then 'w' mode will create a new file.
# read mode 'r' don't create the file.
```

```python
with open("file2.txt",'w') as f:
    f.write("Hello World, Hola Buenos dias")
```

```python
with open("file2.txt",'r') as f:
    print(f.read())
```

```python
# In 'w' mode data get overwrite on previous data
with open("file2.txt",'w') as f:
    f.write("Hello World, How can i help you")
```

```python
with open("file2.txt",'r') as f:
    print(f.read())
```

```python
with open("file2.txt",'a') as f:
    f.write("I am your new Voice agent")
```

```python
with open("file2.txt",'r') as f:
    print(f.read())
```

```python
with open("file2.txt",'a') as f:
    f.writelines(["My agent work with voice which speak with you.", "It works based on speech to speech conversion model"])
```

```python
with open("file2.txt", 'r') as f:
    print(f.read())
```

```python
# w+ mode: write/read mode
# w+ mode will create the previous content 
# w+ create new file if file don't exist

with open("file3.txt", 'w+') as f:
    f.write("This is new file")
    f.seek(0) # act as pointer to read the file.
    print(f.read())
```

```python
# r+ : read/write mode
# File must exist, otherwise shows error

with open("file3.txt", 'r+') as f:
    f.write("File3 is overwritten")
    f.seek(0)
    print(f.read())
```

```python
# a+ : append/ read 

with open("file3.txt", 'a+') as f:
    f.write("a new sentence is updated or appended.")
    f.seek(0)
    print(f.read())
```
