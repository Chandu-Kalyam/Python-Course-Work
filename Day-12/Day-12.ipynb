# Creating different types of tuples

```python
empty_tuple = ()
empty_tuple2 = tuple()

print(empty_tuple, type(empty_tuple))
print(empty_tuple2, type(empty_tuple2))
```

```python
het_tuple = (10, 10.5, "hello", True)
print(het_tuple)
```

```python
dup_tuple = (10, 10, 10, 20, 30)
print(dup_tuple)
```

```python
# list inside a tuple tuple(list())
nested_ = ((1,2,3), [50,50], {50,60,70,80}, "Hello")
nested_tuple = tuple([1,5,5,6])
print(nested_)
print(nested_tuple)
```

```python
# convert set into tuple
set1 = {2,3,5,6}
tuple1 = tuple(set1)
print(tuple1)
```

```python
values= range(1, 11)
tuples2 = tuple(values)
print(tuples2)
```

```python
print(tuple(range(1, 11)))
```

```python
t1 = (1, 2, 4)
t2 = (5,4,2)
t3 = (7,8,9)
t4 = t1 + t2 + t3
print(t4)
```

```python
# tuple comprehension
t5 = tuple(i for i in range(10))
print(t5)
```

# Insert, delete and update operations in directly

```python
tuple1 = (10, 20.5, "Hello", [10, 20, 30])
```

```python
tuple1[3][2] = 5000
```

```python
tuple1
```

```python
# tuple items cannot be updated directly
# tuple1[3][2] = 5000
# complete item in tuple cannot be updated directly
# but if item is mutable then elements inside this mutable item can be updated
```

```python
# insert
tuple3 = (10, 20, 30)
list1  = list(tuple3)
print(list1)
list1.append(500)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
# delete
list3 = list(tuple3)
list3.remove(30)
tuple3 = tuple(list3)
print(tuple3)
```

```python
# update
list3 = list(tuple3)
list3.insert(2,100)
print(tuple(list3))
```

# Popular inbuilt function for tuple

```python
salary = (6,7,100,600,700,90, 10, 80, 120)
print(len(salary))
print(min(salary))
print(max(salary))
print(sum(salary))
# We cannot apply the sort method on tuple salary.sort()
sorted_list = sorted(salary)

print(sorted_list)

# sort() is inplace function
# inplace means sorting is done in the same data structure

# sorted() function return the list
# sort() fucntion don't return anything
```

```python
l1 = [34, 40,50 , 60, 1,3,4, 70]
l1.sort()
print(l1)
```
