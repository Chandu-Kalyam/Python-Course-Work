```python
# Types of ML
# Supervised learning: It works on label data, input + output . Output will be given to model for training.
# Sl has Regression, classification
# Regression: It works on numerical data, it predicts numerical,
# Ex: Linear Regression, MLR, SVR, KNN Regressor, Decision Tree Regression, Ensemble(Random ForestRegression,)
# Classification: It works on categorical data and predicts categories
# Ex: Naive, KNN, SVM, Decision Tree classifier, Logistic Regression.

# Unsupervised learning: It works on unlabeled data, input only no output is given. No output will be given to model for training.
# USL has clustering or grouping
# Ex: K-means, DBSCAN, PCA(Dimensionality Reduction)

# Reinforcement Learning: It works based on feedback + reward, agent + environment.
```

# Simple Linear Regression

```python
# Finding the Best Fit Line
# Best Fit line : which is more nearest to datapoint.
```

```python
# libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model    # used for calling SLR model
from sklearn.model_selection import train_test_split  #
```

```python
df = pd.read_csv("Income_dataset_2025.csv")
df
```

```python
df.info()
```

```python
plt.figure(figsize=(4,3))
plt.scatter(df["Year"], df["Income"])
plt.show()
```

```python
# separating the input feature and output as target
x = df.drop("Income", axis = 1)  # x is input feature / it is dataframe
y = df["Income"]  # Target/ Output label it is series
```

```python
# split data into training and testing
x_train,x_test,y_train,y_test  = train_test_split(x,y,test_size = 0.20,train_size = 0.80, random_state = 42)

print(len(x_train))
print(len(x_test))
```

```python
# train the model
reg_obj = linear_model.LinearRegression() # creating SLR model object
reg_obj.fit(x_train,y_train)  # fit() function is finding the best fit line.
```

```python
reg_obj.predict([[2030]])
```

```python
# 07/11/2026
# Performance of ML Model
reg_obj.score(x_test,y_test)
```

```python
reg_obj.score(x_train,y_train)
```

```python
y_predict = reg_obj.predict(x_test)

# for actual y_train is input, y_test is output, for x_test data we are predicting the value and storing in y_predict
# compare actual vs predicted
for actual, pred in zip(y_test, y_predict):
    print(actual,": ", pred)
```

```python
# plotting the best fit line
plt.scatter(df["Year"], df["Income"])
plt.plot(df["Year"], reg_obj.predict(df[["Year"]]), color = "red")
```
