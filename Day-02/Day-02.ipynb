# Type casting

```python
# Integer typecasting : int()
v1 = int(5.7)  # float to int
v2 = int("300")  # String to int
v3 = int(True)   # bool to int
v4 = int(False)   # bool to int
v5 = int("1111",2)   # binary to int
print(v1, type(v1))
print(v2, type(v2))
print(v3, type(v3))
print(v4, type(v4))
print(v5, type(v5))
```

```python
# float typecasting: float()
f1 = float(10)        # integer to float
f2 = float("4.55")    # String to float
f3 = float(True)      # bool to float

print(f1, type(f1))
print(f2, type(f2))
print(f3, type(f3))
```

```python
# convert to string: str()
s1 = str(10)     # int to string
s2 = str(4.5)
s3 = str(True)
print(s1, type(s1))
print(s2, type(s2))
print(s3, type(s3))
```

```python
# conver to boolean: bool()
b1 = bool(1)
b2 = bool(0)
b3 = bool(1000)
b4 = bool("Hello")
print(b1, type(b1))
print(b2, type(b2))
print(b3, type(b3))
print(b4, type(b4))
```
