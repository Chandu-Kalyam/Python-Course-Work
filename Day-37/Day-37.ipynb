```python
import numpy as np
```

```python
# Normal distributed data: human height data
# loc = mean, scale = Standard deviation
# SD: how much far is the data from the mean
np.random.normal(loc = 5, scale = 0.5, size = 10)
```

```python
np.random.randn(4,2,3)
```

```python
# Used to generate random integer values between specified range
np.random.randint(10,100,5)
```

```python
np.random.randint(10, 100, (3,2,5))
```

```python
# choice(): used to work with categorical data
np.random.seed(42) # if we want to fix the random value then we use seed function.
food_item = np.array(["Holiga", "Lemon_rice", "Payasam", "Pulihora"])
np.random.choice(food_item, size = 3)
```

# create a sequences 

```python
arr1 = np.arange(1, 11)
print(arr1)
print(type(arr1))
```

```python
# shuffle() function is used for shuffle the elements inplace
# This function change the original array means it works inplace
np.random.shuffle(arr1)
print(arr1)
```

```python
# permutation() function is used for shuffle the elements.
# this function don't change the original array.It generate new numpy array
arr2 = np.random.permutation(arr1)
print(arr2)
```

```python
arr3 = np.linspace(1,10,5)  #gap formula=  (10 - 1) / (5 - 1)
print(arr3)
```

```python
arr3 = np.linspace(1,10,4)
print(arr3)
```
