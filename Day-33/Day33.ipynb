# Handling inconsistent data

```python
# Make the column name consistent
# Make all the column name start with capital letter
df.columns = df.columns.str.title()   # str.lower(), str.upper()
```

# Data transformation using apply() & map() function

```python
# apply(): this function is used to execute lambda function
# Lambda function is one line function
# if you want to apply lambda function on DataFrame then we use the apply() function.
```

```python
df1 = df.copy()
```

```python
# On the festival occation: Sai Ram decided to double the employee salary.
# We applied the lambda function on a single column.
df1["Salary"] = df1["Salary"].apply(lambda x: x * 2)
df1["Salary"]
```

```python
# Create a Bonus or No Bonus
# Apply function with complete DataFrame
df1["Diwali_gift"] = df1.apply(lambda row : "Bonus" if(row["Salary"] < 300000 and row["Age"] > 40) else "No_Bonus", axis = 1)
df1
```

# Aggregation and Grouping functions

```python
df.info()
```

```python
# Sum of the total
df["Salary"].sum()
```

```python
# Average of the total salary
# mean = sum of all items / total elements
df["Salary"].mean()
```

```python
# Median : Middle value after sorting the data.
# even : sum the middle values / 2
df["Salary"].median()
```

```python
# Mode: most frequent value
# mode works with categorical data or numerical data with repeated values

df["Profession"].mode()
```

```python
df['Profession'].unique()
```

```python
df["Profession"].value_counts()
```

```python
# Minimum salary
df['Salary'].min()
```

```python
# Maximum salary
df['Salary'].max()
```

```python
# Variance: how much the data far away from mean in squared terms.
df['Salary'].var()
```

```python
# Standard deviation: How much data far away from mean
df["Salary"].std()
```
