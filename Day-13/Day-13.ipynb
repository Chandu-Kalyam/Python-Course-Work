# Creating different types of set

```python
empty_set = set()

set1 = set([10,20,30,40])
set2 = set((10,20,30,40))
# frozenset is immutable (cannot perform inset, delete, update)
fro_set = frozenset([10,20,30,40,50])

#Nested set
nested_set = {(10,20,30),"Hello", 30, fro_set} # tuple + frozenset can be store in the set
# nested_set1 = {[10,20,30],"Hello", 30}  list cannot be stored inside the set
# nested_set1 = {{10,20,30},"Hello", 30}  list cannot be stored inside the set

# IMP: Mutable items cannot be stored inside the set.
# Only immutable items can be stored in the set.

print(empty_set)
print(set1)
print(set2)
print(fro_set)
print(nested_set)
```

# Insert in set (add(), update())

```python
set1 = {10, 20.5, "Hello"}

# add() function is used to insert one item in the set
set1.add("world")
set1.add((10,20,30,40))
print(set1)
```

```python
# update() is used for inserting multiple items together
set1.update({(1000, 2000)})
print(set1)
```

```python
# Multiple items can be passed to update() function in a set or tuple or list 
s1 = {101, 102}
t1 = (201, 202)
l1 = [301, 302]
set1.update(s1)
set1.update(l1)
print(set1)
```

# delete operation on set (remove(), discard(), pop(), )

```python
dinner_set = {"Bagara rice", "Biryani", "Papu", "Chapathi", "Fish","Mutton", "prawns"}
# remove() function delete the one items whose name is provided
dinner_set.remove("prawns")
# remove() will raise on ERROR if item don't exist
# discard() function delete the one item whose name is provided
dinner_set.discard("Biryani")

print(dinner_set)
```

```python
# discard do not raise ERROR if item is not present
dinner_set.discard("prawns")
print(dinner_set)
```

```python
# pop() will delete any random item
dinner_set.pop()
print(dinner_set)
```

```python
dinner_set.clear()
print(dinner_set)
```

# Apply popular inbuilt functions on set

# containment operation on set

```python
south_festivals = {"Sankranthi", "Bhogi", "Ugadi", "Onam", "Dussahra"}
if "Holi" in south_festivals:
    print("Happy Holi")
elif "Sankranthi" not in south_festivals:
    print("Happy Sankrathi")
elif "Ugadi" in south_festivals:
    print("Happy Ugadi")
```
