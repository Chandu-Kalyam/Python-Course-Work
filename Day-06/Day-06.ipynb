# match case

```python
# Match case is used to replace multiple if else condition 
# Match case is most powerful than the Switch case statement
# Match case works with variable and data structure.
```

```python
# Example 1:
num = int(input("Enter a number:"))
match num:
    case 10:
        print("You entered 10")
    case 20:
        print("You entered 20")
    case 30:
        print("You entered 30")
    case _:   # this is default same like switch statement
        print("Something else")
```

```python
biryani = input("Enter famous hyderabadi biryani: ")
match biryani:
    case "Paradise":
        print("Spicy and taste")
    case "Shahdab":
        print("double masala")
    case "road side":
        print("Budget friendly")
    case _:
        print("Pg biryani")
```

```python
# Example 3: passing data sturcture to match case 

data = [1, 2, 3, 4]
match data:
    case [4, 3, 2, 1]:
        print("Hello")
    case [1, 2, 3, 4]:
        print("world")
```

```python
# Example 3: if condition with match case
val = int(input("Enter a number: "))

match val:
    case x if x > 0:
        print("It is a positive number")
    case y if y < 0:
        print("It is a Negative number")
    case _:
        print("It is a Zero")
```

```python
# Example greatest of three numbers
a,b,c = map(int, input("Enter a three numbers: ").split())
match a,b,c:
    case x,y,z if x>y and x>z:
        print(x)
    case x,y,z if y>x and y>z:
        print(y)
    case _:
        print(z)
```
