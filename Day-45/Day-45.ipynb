```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("tips")
plt.figure(figsize= (4,3))
sns.boxplot(y = "tip", data = df)
# Use: boxplot is used for finding the outlier.
```

```python
df.columns
```

```python
sns.boxplot(y = "total_bill", data = df)
```

```python
sns.boxplot(x = "day", y = "total_bill", data = df)
```

```python
# finding daywise outliers in tip
sns.boxplot(x = "day", y = "tip", data = df,  hue = "smoker")
```

# Density plot (KDE plot)

```python
# KDE: Kernal Density Estimate
# Use: Density plot is used to visualize probability of a continuous data.
plt.figure(figsize = (4,3))
sns.kdeplot(data = df, x = "total_bill", fill = True)
```

# Violin plot

```python
# violin plot = Box plot and density plot
plt.figure(figsize = (4,3))
sns.violinplot(x = "day", y = "total_bill", data = df, )
```

# Regression plot

```python
# Use: to plot the linear regression line
plt.figure(figsize = (4,3))
sns.regplot(x = "total_bill", y = "tip", data = df)
```

# FacetGrid

```python
# Use: Create multiple plots based on classes in categorical variable
plt.figure(figsize = (4,3))
g = sns.FacetGrid(df, col = "day")
g.map_dataframe(sns.scatterplot, x = "total_bill", y = "tip")
```

```python
plt.figure(figsize = (4,3))
g = sns.FacetGrid(df, col = "day")
g.map_dataframe(sns.lineplot, x = "total_bill", y = "tip")
```

# Joint plot

```python
# Use: combine 2 plots: Histogram + scatterplot
plt.figure(figsize = (4,3))
sns.jointplot(data = df, x = "total_bill", y = "tip", kind = "scatter")
```
