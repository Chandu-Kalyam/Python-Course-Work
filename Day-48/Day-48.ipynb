# Logistic Regression

```python
# Logistic Regression is mainly for classification problems.
# Logistic regression is mainly used for Binary classification
```

```python
# libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("insurance_logistic_dataset_10k.csv")
df
```

```python
df["bought_insurance"].unique()
```

```python
plt.figure(figsize = (4,3))
plt.scatter(df["age"], df["bought_insurance"])

# The shape of logistic regression model is S
```

```python
sns.lineplot(x = "age", y = "bought_insurance", data = df)
```

```python
# split the data into input feature and target 
x = df.drop("bought_insurance", axis = 1)
y = df["bought_insurance"]
```

```python
# divide the data into training and testing
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.2,random_state = 42)
```

```python
# train the logistic regression model
model = LogisticRegression()
model.fit(x_train, y_train)
```

```python
# performance of model
model.score(x_test, y_test)
```

```python
# actual vs predicted
# where the actual test output is stored
actual = y_test[:10]
y_predict = model.predict(x_test[0:10])

for act, pred in zip(actual,y_predict):
    print(f"{act} : {pred}")
```
