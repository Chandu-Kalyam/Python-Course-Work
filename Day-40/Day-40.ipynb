# Bar chart

```python
# USE: used to shows the discrete values of categorical data.
food_items = ["Idli", "Dosa", "Vada", "Sambar", "Puri"]
like_count = [2,7,3, 10,12]

plt.figure(figsize=(4,3))

plt.bar(food_items, like_count, color=["orange", "blue", "purple", "grey", "black"])

for i,count in enumerate(like_count):
    plt.text(i, count, count, ha = "center") # plt.text(x_index, y_index, value, position)

plt.xlabel("Food_Items")
plt.ylabel("No of likes")
plt.title("Bar Chart")
```

# Horizontal barchart

```python
# USE: when the categorical data values are very large
food_items = ["Idli", "Dosa", "Vada", "Sambar", "Puri", "PaneerLemonSweetYummyParathaRice"]
like_count = [2,7,3, 10,12, 9]
plt.figure(figsize=(4,3))
plt.barh(food_items, like_count)
```

# Histogram

```python
# USE: Historgram works on continuous data
# Histogram is popularly used to find the shape of the data.

import numpy as np

np.random.seed(42)
data = np.random.randn(1000) # normal distributed data: range = -3 to +3 ==> 99%
data

plt.figure(figsize=(5,5))

plt.hist(data, edgecolor = "black", bins = 5)

plt.title("Histogram")
plt.xlabel("X-data")
plt.ylabel("y-data")
```

```python
# Interview question: what is the difference between bar chart and histogram?
# Ans: Bar chart works on categorical data values while histogram works on continuous data
# Bins in histogram don't have space between them.
```

# Line chart

```python
# USE: to show trends over the time
# time series data

import pandas as pd
dates = pd.date_range(start = "2026-01-01", periods = 12, freq = "ME")
dates
```

```python
icecream_sales = [60, 70, 100, 20, 40, 50, 90, 120, 130, 140, 170, 190]
len(icecream_sales)
```

```python
plt.figure(figsize=(5,5))
plt.plot(dates, icecream_sales, marker = "o" )
for d,ice in zip(dates, icecream_sales):
    plt.text(d, ice + 5, ice, ha = "center")  # plt.text(x_index, y_index, value, position)

plt.xticks(dates, dates.strftime("%b"), rotation = 45)
plt.show()
```

# Multiline chart

```python
# USE: to compare multiple trends side by side

dates = pd.date_range(start = "2016-01-01", periods = 10, freq = "YE")
dates
```

```python
Male_expense = [6, 10, 6.5, 4.5, 7, 8, 10, 12,5, 13]
Female_expense = [14, 15, 0, 18, 20, 25, 40, 16, 30, 35]


plt.figure(figsize = (5,3))
plt.plot(dates, Male_expense, label = "Boys")
plt.plot(dates, Female_expense, label = "Girls")
plt.legend()
```
