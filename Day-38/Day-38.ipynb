```python
import numpy as np
```

### Array shape, reshape and trasnpose

```python
food_arr = np.array([["Papu", "Rice", "Dosa"],["Sugarcane", "coconut", "mangoshake"]])
print(food_arr)
print(type(food_arr))
print(food_arr.ndim)
print(food_arr.shape)
print(food_arr.transpose())
```

```python
# reshpae(): reshaping the array
arr1 = food_arr.reshape(3,2)
arr1
print(arr1.ndim)
```

```python
arr2 = food_arr.reshape(6)
print(arr2)
print(arr2.ndim)
```

```python
# converting 2D to 3D
arr3 = food_arr.reshape(2,1,3)
print(arr3)
print(arr3.ndim)
```

```python
# we can convert anyD array into any other D array
# We Identify that the size of input array must be same as output array
```

```python
arr3 = food_arr.reshape(1,2,1,3)
print(arr3)
print(arr3.ndim)
```

```python
# ravel(): convert anyD array to 1D array
food_arr = np.array([[["Papu", "Rice", "Dosa"],["Sugarcane", "coconut", "mangoshake"]]])
food_arr.ravel()
```

```python
# transpose: row <--> column
food_arr = np.array([["Papu", "Rice", "Dosa"],["Sugarcane", "coconut", "mangoshake"]])
print(food_arr.T)
print()
print(food_arr.transpose())
```

### Indexing, slicing and modification in numpy array

```python
sweets_array = np.array([["Potharekalu", "Jangiri", "palakova"], ["firni", "Ghevar", "Gulab jamun"]])
print(sweets_array[0][2])
```

```python
print(sweets_array[1,1])   # [row_no, col_no]
```

```python
print(sweets_array[0,1:])   # 0: ROW, 1: col range
```

```python
print(sweets_array[1,:2])  
```

```python
print(sweets_array[:,0])  
```

```python
print(sweets_array[:,1])  # all rows and 1st column
```

```python
print(sweets_array[-1,-2])
```

```python
print(sweets_array[-1:,-1:-2:-1])   # ROW: -1 to end, end = -2-(-1) = -2 + 1 = -1
```

# Modification, insert, delete on numpy array

```python
# Modification can be done inplace on numpy array
# Insertion and deletion cannot be performed inplace
# numpy array take the fix space in memory
```

```python
south_heroine = np.array([["No_one","Kajal", "Tamana", "Samantha"],["Imli", "Zena", "Nura", "Pooja"]])

south_heroine[1,0] = "Emily"
print(south_heroine)
```

```python
south_heroine[:,0] = ["Chinky","Minki"]
print(south_heroine)
```

```python
south_heroine[1,1:3] = ["Tinki","Pinki"]
print(south_heroine)
```

```python
# np.insert(): inserting element inside the numpy array
sweets_array = np.array([["Potharekalu", "Jangiri", "palakova"], ["firni", "Ghevar", "Gulab jamun"]])

# axis = 0 : insert new row
new_sweets = np.insert(sweets_array, 2, ["Laddu", "kajuBurfi", "Rasamalai"], axis = 0)
print(new_sweets)
```

```python
# axis = 1: insert new column
new_sweets = np.insert(sweets_array, 0, ["A", "B"], axis = 1)
print(new_sweets)
```

```python
# Deleting axis = 0 means deleting row
new_arr = np.delete(sweets_array, 0, axis = 0)
print(new_arr)
```

```python
# axis = 1 means deleting column
new_arr = np.delete(sweets_array, 1, axis = 1)
print(new_arr)
```
