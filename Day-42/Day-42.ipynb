# plt.subplot(nrows, ncols, index)

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
digits = load_digits()
```

```python
mobile_company = ["Realme", "Samsung", "Nokia", "Lava", "Apple", "IQoo", "OnePlus"]
market_share = [15, 25, 2, 3, 30, 5, 20]

food_items = ["Idli", "Dosa", "Vada", "Sambhar", "Puri"]
like_count = [2, 7, 3, 10, 12]

plt.figure(figsize = (6,4))
plt.subplot(2,3,1)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,2)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,3)
plt.imshow(digits.images[4])

plt.subplot(2,3,4)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])

plt.subplot(2,3,5)
plt.pie(market_share, labels = mobile_company)

plt.subplot(2,3,6)
plt.barh(food_items, like_count, color = ["orange", "blue", "purple", "grey", "black"])
plt.tight_layout()  # use this if subplots are overlapping
plt.show()
```

```python
Dinner_dream_thali=["Biryani", "Chicken", "Rusgulla", "Roti", "Lassi"]
prec=[30,20,10,20,20]

_, axes= plt.subplots(2,2,figsize=(10,10))  # (nrows, ncols, full image size)
titles=["Gopi","Hari","Ajay","Dinesh"]
print(axes)
axes_id=axes.ravel()
print(axes_id)


for titles, ax in zip(titles,axes.ravel()):
    ax.pie(prec, labels=Dinner_dream_thali, autopct='%1.1f%%')
    ax.set_title(titles)

plt.tight_layout()
plt.show()
```
