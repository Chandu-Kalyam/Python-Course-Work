# Creating different types of list

```python
empty_list = []
empty_list2 = list()

print(empty_list, type(empty_list))
print(empty_list2, type(empty_list))
```

```python
Het_list = [10, 20.5, "Hello", True]
print(Het_list)

dup_list = [10, 10, 10, 10]
print(dup_list)
```

```python
# Nested list
nest_list = [[i,j] for j in range(3) for i in range(3)]
print(nest_list)
```

```python
print(nest_list[1][0])
```

```python
concat_list = Het_list + dup_list + nest_list
print(concat_list)
```

# Updating list items

```python
list_politician = ["Jagan Mohan Reddy", "Vijay", "KCR"]

list_politician[1] = "Dance with"
print(list_politician)
```

```python
# Update list elements with More
list_politician[1:4] = ["Rahul", "Kejriwal","Yogi", "Mamta","Anushka"]
print(list_politician)
```

```python
list_politician[-2] = "Dance with"
print(list_politician)
```

```python
list_politician[-5: -1] = ["Dance with"]
print(list_politician)
```

# insert items in list

```python
# append(), extend(), insert()
```

```python
l2 = []
# append() function is used to insert the element at the last of the list
l2.append(10)
l2.append(20)
l2.append(30)
print(l2)
```

```python
# extend() : this function is used to insert multiple elements in the last
l2.extend([40,50,60,70,80,90,100])
print(l2)
```

```python
# insert() can be used to insert at specific index
l2.insert(4,"Dinesh")
l2
```
