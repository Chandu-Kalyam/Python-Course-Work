```python
# the word polymorphism means "many forms"
# In polymorphism, same function or same operator can perform many different tasks.
# Type of polymorphism:
#  1. Function overloading
#  2. Operator overloading
#  3. Method overloading
```

### 1. Function overloading

#### using default arguments for function overloading

```python
# Example-1:
def fun(var1, var2 = None, var3 = None):
    if var2 is not None and var3 is not None:
        print(var1, var2, var3)
    elif var2 is not None:
        print(var1, var2)
    else:
        print(var1)
     
    
fun(10)
fun(10, 20)
fun(10,20,30)
```

```python
# Example-2: default argument
def cal(a=0, b = 0, c = 0, d = 0):
    return a+b+c+d

print(cal())
print(cal(100))
print(cal(100,200))
print(cal(100,200, 300))
print(cal(100,200,300, 400))
```

## *args for function overloading

```python
def cal(*args):
    return sum(args)

print(cal())
print(cal(100))
print(cal(100,200))
print(cal(100,200, 300))
print(cal(100,200,300, 400))
print(cal(100,200,300,400,500,600,700,800,900))
```

```python
# function overloading using decorator
from multipledispatch import dispatch

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int, int)
def fun(var1, var2):
    print(var1, var2)
    
@dispatch(float, float, float)
def fun(var1, var2, var3):
    print(var1, var2, var3)
    
fun(10)
fun(10, 20)
fun(10.4,10.5,10.5)
```

# Operator overloading

```python
# we will perform different operations on the user defined objects using operators.
```

```python
class Dosa:
    def __init__(self, variety, price):
        self.variety = variety
        self.price = price
        
    def __add__(self,d2_val):
        return self.price + d2_val.price
    
    def __sub__(self, d2):
        return d2.price - self.price
    
    def __mul__(self,d2):
        return self.price * d2.price
    
    def __truediv__(self,d2):
        return d2.price / self.price
    
    def __floordiv__(self,d2):
        return d2.price // self.price
    
    def __mod__(self, d2):
        return self.price % d2.price
    
    
    
d1 = Dosa("Masala", 60)
d2 = Dosa("Icecream", 110)


print(d1 + d2)
print(d1 - d2)
print(d1 * d2)
print(d1 / d2)
print(d1 // d2)
print(d1 % d2)
# print(d1 ** d2) // __pow__
```

```python
# comparison operators overloading

class Dosa:
    def __init__(self, variety, price):
        self.variety = variety
        self.price = price
        
    def __gt__(self,d2):
        return self.price > d2.price
    
    def __lt__(self,d2):
        return self.price < d2.price
    
    def __ge__(self,d2):
        return self.price >= d2.price
    
    def __le__(self,d2):
        return self.price <= d2.price
    
    def __eq__(self,d2):
        return self.price == d2.price
d1 = Dosa('Masala', 60)
d2 = Dosa("Icecream", 110)

print(d1 > d2)    # __gt__()
print(d1 < d2)    # __lt__()
print(d1 >= d2)   # __ge__()
print(d1 <= d2)   # __le__()
print(d1 == d2)
```
