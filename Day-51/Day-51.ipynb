# K Nearest Neighbour (KNN)

```python
# KNN is a supervised ML algorithm
# KNN mainly used for classification
# Although KNN can also be used for Regresssion based problems.
# KNN can be used for binary or multiclass classification
# KNN is also known as lazy mL algorithm.
# Maximum calculation are done at the time of prediction.
```

```python
# working principal of KNN:
# In KNN, we look for K nearest data points.
# The data points which are in majority decide the category.
# Its good idea to take k value as odd value. so there won't be tie.
# KNN is slow because this is time consuming to find the nearest data points.
```

```python
# libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
```

```python
df = pd.read_csv("teleCust1000t.csv")
df
```

```python
x = df.drop("custcat", axis = 1)
y = df.custcat
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state=42)
```

```python
# scale the training and testing data
scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# train the model
model = KNeighborsClassifier(n_neighbors=85)
model.fit(x_train_scaled,y_train)
```

```python
model.score(x_test_scaled,y_test)
```

# apply Naive bayes

```python
from sklearn.naive_bayes import GaussianNB
model1 = GaussianNB()
model1.fit(x_train_scaled,y_train)
```

```python
model.score(x_test_scaled, y_test)
```

# apply svm

```python
from sklearn.svm import SVC
model2 = SVC(kernel = "")
model2.fit(x_train_scaled,y_train)
```

```python
model2.score(x_test_scaled,y_test)
```

```python
df2 = pd.read_csv("Chinky_shopping_regression_dataset.csv")
df2
```
