# Decision Tree

```python
# Decision Tree is a Supervised ML algorithm
# Decision tree is mainly used for classification problems.
# Decision tree can also be used for Regression based problems.
# Decision tree algorithm take time while training.
# Decision tree is fast while making predictions

# Features: Age, Weight, Smoker,  Target: health issue
```

```python
# Libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_wine
```

```python
wine_dataset = load_wine()
dir(wine_dataset)
```

```python
print(wine_dataset.DESCR)
```

```python
x = pd.DataFrame(wine_dataset.data, columns = wine_dataset.feature_names)
y = wine_dataset.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.25, random_state = 42)
```

```python
x_train
```

```python
# train the model
# max_depth helping in model Pruning
model = DecisionTreeClassifier(max_depth = 5, random_state = 42)
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
# plotting the DT
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

plot_tree(model, feature_names = wine_dataset.feature_names,class_names = wine_dataset.target_names, filled = True)
plt.title("Decision Tree for wine dataset")
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.datasets import fetch_california_housing
```

```python
housing = fetch_california_housing()
dir(housing)
```

```python
print(housing.DESCR)
```

```python
x = housing.data
y = housing.target
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size = 0.25, random_state = 42)
```

```python
model1 = DecisionTreeRegressor(max_depth = 3)
model1.fit(x_train,y_train)
```

```python
model1.score(x_test,y_test)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree
plt.figure(figsize = (10,5))
plot_tree(model1, feature_names = housing.feature_names,class_names = housing.target_names, filled = True, fontsize = 6)
plt.title("Decision Tree for house dataset")
```
