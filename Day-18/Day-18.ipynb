```python
# Example-1:
# class is a keyword, food is the class name
# class variable: the value of class variable remains same for all the objects.
# class variable value cannot be change by object.
# class variable value can be only change by class name.


class Food:  # all classname must be in CamelCase
    FOOD_NAME = "Biryani"    # class variable
 


obj = Food()    # creating the object (obj) of Food() class
print(obj.FOOD_NAME)  # accessing the class variable

# access the class variable with class name
print(Food.FOOD_NAME)


# class variable value can be only change by class name.
Food.FOOD_NAME = 'Egg'

obj1 = Food()
print(obj1.FOOD_NAME)   # class object (obj2) can access the class data.
print(Food.FOOD_NAME)

print()
# class variable value cannot be change by object.
obj1.FOOD_NAME = "Panner"    # here obj1, is creating its own local variable with same name. 
print("We can only change value using with obj inside but u cannot change the main class variable like Food.FOOD_NAME")
print(obj1.FOOD_NAME)
print(Food.FOOD_NAME)
```

```python
# Example-2: constructor
# when ever we create a obj the constructor is automatically called by itself
# constructor function automatically get called when object is created.
# constructor function is used for initializing the variable, while creating the new object.

# instance variable value is different for every object.
# class variable value is same for all the objects.
# self is a variable which points to the current object.

class Bike:
    NO_TYRE = 2          # class variable
    
    def __init__(self, name, price):   # class constructor
        print("I am constructor function")
        self.name = name        # instance variable
        self.price = price
        print(f"Vicky has {self.name} which costs around {self.price} Lakh")
        
obj = Bike("RE-Classic-350", 2.50)
print(obj.name)
print()
obj1 = Bike("Yamaha-FZ",1.7)
print(obj1.name)
```

```python
# Example-3: member function/ methods
# Member function dont get called automatically, we need to call them ourself.

class Bike:
    NO_TYRE = 2          # class variable
    
    def __init__(self, name, price):   # class constructor
        print("I am constructor function")
        self.name = name        # instance variable
        self.price = price
        print(self.name, self.price)
        
    def display(self):
        print(f"Vicky has {self.name} which costs around {self.price} Lakh")
        
obj = Bike("RE-Classic-350", 2.50)
obj.display()
print()
obj1 = Bike("Yamaha-FZ",1.7)
obj1.display()
```
