# for loop 

```python
animals = ['Puli', 'Pili', 'Kukka', 'Pandi', 'Kothi']
for animal in animals:
    print(animal)
```

```python
food_name = "Chicken_Dum_Masala_Biryani"
for food in food_name:
    print(food, end=" ")
```

```python
# for loop with range()
# range() function is used to generate list of elements
# syntax of range : range(start, stop, step)
# start = 0    by default
# end = end - 1
# step = 1     by default
for i in range(0,12):
    print(i, end=" ")
```

```python
# for loop with range()
# range() function is used to generate list of elements
# syntax of range : range(start, stop, step)
# start = 0    by default
# end = end - 1
# step = 1     by default
for i in range(0,12,2):
    print(i, end=" ")
```

```python
# Example -4: break statement
# break statement will bring the control out of the loop
for num in range(1,6):
    if num == 3:
        break
    print(num)
```

```python
# Example - 5: Continue statement
# Continue statement take the control to the start of the loop

for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# Example - 6: nested loop
# for every value of outer loop, inner loop will completely execute.
for i in range(3, 6):  # n = 3 times
    for j in range(2, 5): # m = 3 times           here time complexity about O(n * m )
        print(i,j)
```

```python
# Example - 7: else statement
# else statement executes if loop complete execution successfully

for i in range(5):
    if i == 2:
        continue
    elif i == 40:
        break
    else:
        print(i)
    pass
    
else:
    print("Wow u have finished the loop")
```

```python
n = int(input("enter a number:"))
for i in range(n, 0, -1):
    print("* " * i)
```

```python
n = int(input("enter a number:"))
for i in range(n , 0, -1):
    print(" " * i + "* " * (n - i+1))
```
