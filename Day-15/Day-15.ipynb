```python
#  example-3: nested function: function inside function
def outer():
    print("I am outer function")
    def inner():
        print("I am inner function")
        
    inner()     # calling from inside the building function
        
outer()
```

```python
def addition(a, b):
    return a + b
def sub(a, b):
    return a - b
def mul(a, b):
    return a * b
def div(a, b):
    return a/b
def rem(a, b):
    return a%b

user_input = input("Enter which type of operation to be performed on numbers: ").lower()
num1, num2 = map(int, input("Enter two numbers: ").split())
if user_input == 'add':
    print(f"Addition of {num1}, {num2}: {addition(num1, num2)}")
if user_input == 'sub':
    print(f"Subtraction of {num1}, {num2}: {sub(num1, num2)}")
if user_input == 'mul':
    print(f"Multiplication of {num1}, {num2}: {mul(num1, num2)}")
if user_input == 'div':
    print(f"Division of {num1}, {num2}: {div(num1, num2)}")
if user_input == 'rem':
    print(f"Addition of {num1}, {num2}: {rem(num1, num2)}")
```

# Default parameters

```python
def fun(var=100):
    print(var)
    
fun()
```

```python

def func(*args):
    result = 0
    for num in args:
        result += num
    return result
    

total = func(1, 3, 5, 5 ,2)
print(total)
```

```python
def func2(x = 10, y = 20):
    print(x+y)
    
func2()
func2(100)
func2(100, 200)
```

# passing arbitary number of arguments using *args

```python
def anyarg(*args):
    for num in args:
        print(num)
        
anyarg()
anyarg("Raj")
anyarg("R", "G", "B")
```

# Passing arbitary Key-value based argument using **kargs

```python
# Passing Key-value based argument using **kargs
def keyvalargs(**kargs):
    for key, value in kargs.items():
        print(key, value)
        
keyvalargs(name = "Raj", age = 20)
```
