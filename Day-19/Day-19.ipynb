# Encapsulation

```python
# Def: Encapsulation refers to the binding of data(attributes) and functions into one unit(class).
# Encapsulation restrict the direct access of the data.
# we can access the variable/attribute/data by using object outside the class.
# we can access the variable/attribute/data by functions defined inside the class.
```

```python
class Bike:
    NO_TYRE = 2          # class variable/ attribute/ data
    
    def __init__(self, name, price):   # class constructor
        print("I am constructor function")
        self.name = name        # instance variable
        self.price = price
    
        
    def display(self):
        print(f"{self.NO_TYRE}, {self.name}, {self.price}")
        
obj = Bike("RE-Classic-350", 2.50)
obj.display()
print()
obj1 = Bike("Yamaha-FZ",1.7)
obj1.display()
```

### Access specifier/ type of attributes in encapsulation

```python
# 1. Public : These variables can be accessed outside the class.
# public variables dont start with _(underscore).
# 2. Protected : These variables can be accessed only inside/child class.
#               Protected variable not Intended to access outside the class.
# protected variables start with SINGLE underscore.

# 3. Private : These variables can be accessed only inside the class.
# Not Intended to access outside or child class.
# Private variable starts with __(double underscore).
```

```python
class Marvels:
    
    def __init__(self, name, power, weapon):
        self.name = name         # public instance variable
        self._power = power      # protected instance variable
        self.__weapon = weapon     # private instance variable
        
    def show(self):
        print(self.name, self._power, self.__weapon)
        
obj = Marvels("Thor", 1000, "Hameer")
obj.show()
print(obj.name)
print(obj._Marvels__weapon)      # obj._classname__privatevariable

# Protected variables can be accessed outside the class, but you should not access.
# In python public, protected, private concept is used just for Sign/ warning. 
# print(obj._power)    # we should not access protected variables even if it is possible feel it is like a warning.
# print(obj.__weapon)  # Attribute error for private variable we cannot access it.
```
