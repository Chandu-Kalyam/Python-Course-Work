# Handling Null Values

```python
# Null: The information which is not provided
# NaN, na
# Null vs 0
# 0 is a value, while Null means no value available.
```

```python
# identify the null values in dataframe
df.isnull().sum()
```

```python
df.isna().sum()
```

```python
# find the null values in a specific column
df[df["Age"].isnull()]
```

```python
# print 
df["Age"].isnull().sum()
```

```python
# Delete the records which is having null values
df1 = df.copy()
df1.dropna(inplace = True)
```

```python
df1.isnull().sum()
```

```python
df1.shape
```

```python
# Why we need to handle the null values
# Ans: inbuild functions might not work properly (numpy and python)
# Pandas function can work but will give the wrong answer as it will not include the NaN value.

import numpy as np
vals = [10, 20, 30, 40, np.nan]
np.mean(vals)
```

```python
df['Age'].mean()
```

```python
# series
ser1 = pd.Series(
    vals
)
ser1.mean()
```

```python
# Approach-1: delete the records which is having null values
# Approach-2: delete the columns which is having the null values
df2 = df.copy()
```

```python
df2.isnull().sum()
```

```python
# axis =1: delete the column which is having null value
# axis = 0: delete the row which is having null value
# how = "any" : delete the column even that have single null value
# how = "all" : delete the column only if all the values in that column is NULL
df2 = df2.dropna(axis = 1, how = "all")
df2.isnull().sum()
```

```python
# Approach-3: replace the null value with same constant
df3 = df.copy()
```

```python
# fillna() function is used to fill the null values with some other value
print(df3["Age"].mean())
df3["Age"] = df3["Age"].fillna(df3["Age"].mean())
```

```python
df3.isnull().sum()
```

```python
df3.shape
```

# Handle Duplicate values

```python
# Identify the duplicate records
df.duplicated().sum()
```

```python
# show the duplicate records 
# if a record is completely matching with same other record
df[df.duplicated()]
```

```python
# keep = False: show both the records
# - ``first`` : Mark duplicates as ``True`` except for the first occurrence.
# - ``last`` : Mark duplicates as ``True`` except for the last occurrence.
# - False : Mark all duplicates as ``True``
df[df.duplicated(keep = False)]
```

```python
df[df.duplicated(keep = "last")]
```

```python
df[df.duplicated(keep = "first")]
```

```python
# Deleting the duplicate records
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```
