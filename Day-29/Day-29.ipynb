## Data Structure in Pandas (Series, Dataframe)

```python
# Series: 1D array
# Dataframe: 2D array
```

# Series Data Structure

```python
# Series is a 1D Homogenous array.
```

```python
ser1 = pd.Series()     # empty series
print(ser1)
print(type(ser1))
```

```python
ser2 = pd.Series([10, 20, 30, 40, 50])
print(ser2)
print(type(ser2))

# strings in pandas are represented as "object".
# Internally, pandas convert all the elements to string if even a single element passed is string.
```

```python
ser3 = pd.Series(["Banana", "Orange", "Guava", "cucumber"])
print(ser3)
```

```python
# pandas series don't support negative indexing.
# ser3[-1]  valueerror
```

# DataFrame : 2D Array

```python
df1 = pd.DataFrame(["Brinjal", "Carrot", "Lady's finger"])     # we created empty data frame.
print(df1)
```

```python
# 2 columns dataframe
df2 = pd.DataFrame(
    {
        "Vegetable": ["Brinjal", "Ladyfinger", "Carrot"],
        "Price" : [50, 20, 30]
    }
)
print(df2)
```

```python
# Convert dictionary to dataframe
fav_hero = {
    "Hero_name" : ["Gopichand", "Pavan Kalyan", "Allu Arjun", "Ram Charan"],
    "Fee" : [20, 0, 100, 50]
}
df4 = pd.DataFrame(fav_hero)
print(df4)
```
