### Delete item from list (remove(), pop(), clear(), del()) 

```python
food_list = ['Papu','Biryani','Fried rice', "Sambar", "Payasam","Curd rice"]
food_list
```

```python
# remove(): delete the item with given name
food_list.remove("Fried rice")
food_list
```

```python
# pop() is used to delete the item from a particular index, pop() without index will delete the list element
food_list.pop()         # pop(we can give the index)
food_list
```

```python
# delete multiple elements together
food_list = ['Papu','Biryani','Fried rice', "Sambar", "Payasam","Curd rice"]
del food_list[2]
del food_list[1:3]
```

```python
food_list
```

```python
# clear() is used to delete all the items from the list 
food_list.clear()
print(food_list)
```

```python
del food_list
```

# Popular list in built functions

```python
Age_list = [22, 33, 22, 20, 19, 18, 21, 16, 19, 20, 99]
print(Age_list)
print(len(Age_list))  
print(min(Age_list))
print(max(Age_list))
print(sum(Age_list))

Age_list.reverse()
print(Age_list)
# In python sort() function use the TIM sort 
# Tim => (Merge + Insert)
# Tim sort => O(n logn)

Age_list.sort()
print(Age_list)

Age_list.sort(reverse=True)
print(Age_list)
```

# list comprehension

```python
# Creating the list in a single line using a loop
nums = [num for num in range(10)]
print(nums)
```

```python
squares = [num**2 for num in range(1, 11)]
print(squares)
```

```python
even = [num for num in range(1, 11) if num%2==0]
print(even)
```

```python
l3 = [i for i in range(1,11) if i%2==0 if i%3==0]
print(l3)
```

```python
l4 = [num if num%2==0 else '' for num in range(1,11)]
print(l4)
```

```python
square_ = [num**2 if num%2==0 else num+2 for num in range(1, 11) if num>5 and num<9]     # Highest priority for right side
print(square_)
```

```python
l5 = [[i for i in range(1, 5)] for j in range(1,3)]
print(l5)
```

```python
l6 = [[[i for i in range(1, 5)] for j in range(1,3)] for k in range(1,3)]
print(l6)
```
