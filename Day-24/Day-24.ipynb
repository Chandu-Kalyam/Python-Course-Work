```python
# Def: EH allows to respond to errot without crashing the running application.
# with EH, we can give our own error message
```

```python
# type of blocks in Exception handling:
# 1. try : put that code where error can occur
# 2. except: handles the error
# 3. else: put code which you want to execute when their is NO error.
# finally: Execute either error or No error.
```

```python
# Example - 1: structure of try-except

try:
    x = 10
    print(x)
except:    # this will catch any type of error
    print("Some error occured...")
    
print("Hello")
a,b = 10, 20
print(a + b)
```

```python
# Example - 2: Exception as e

try:
    print(c)
except Exception as e:
    print(f"Compiler error: {e}")
    print(type(e))
    
print("Hello")
a,b = 10, 20
print(a + b)
```

```python
# Example - 3 : else block: execute if NO error or when try block executes successfully.

try:
    c = 0
    print(c)
except Exception as e:
    print(f"Compiler error: {e}")
    print(type(e))
else:
    print("Hurray, we successfully executed code")
```

```python
# Example - 4 : finally block: execute if NO error or Error it will always executes for anything

try:
    c = 0
    print(c)
except Exception as e:
    print(f"Compiler error: {e}")
    print(type(e))
else:
    print("Hurray, we successfully executed code")
finally:
    print("I will execute whatever happen")
```

### Type of errors:

```python
# NameError : Variable not defined
# ZeroDivisionError: 
try:
    z = 10
    print(z/0)
except NameError:
    print("This is Name Error")
    
# except ZeroDivisionError:
#     print("This is zero division erro")
    
except (NameError, ZeroDivisionError) as e:
    print(f"this have both NameError and ZDError {e}")
except:     
    print("Some Error occur")
```

```python
# IndexError
try:
    l1 = [10, 20, 30]
    print(l1[5])
except NameError:
    print("this is NameError")
except ZeroDivisionError:
    print("This is ZeroDivision Error")
except IndexError:
    print("This is Index Error")
except Exception as e:
    print(type(e), e)
```

```python
# Key Error
try:
    d1 = {"name": "Pavan"}
    print(d1["age"]) 
except KeyError as e:
    print(f"this is key error {e}")
```

```python
# ValueError
try:
    num = int("abc")
    
except ValueError as e:
    print(f"This is a value error {e}")
```
