### Operations on strings

##### 1. Concatenation (+)

```python
str1 = 'Babai '
str2 = 'likes '
str3 = 'Pinni'
print(str1 + str2 + str3)
```

##### 2. Repetition (*)

```python
str6 = "bla "
print(str6 * 10)
```

##### 3. Membership  (in, not in)

```python
str7  = "HelloWorld"
print('Helll' in str7)
print('Helll' not in str7)
print('oW' not in str7)
print("Helo" not in str7)
```

##### 4. Indexing and Slicing

```python
# string[start:end:step]
# start = 0
# end = n - 1
# step = 
Heroine = "Samantha"
print(Heroine[4])
print(Heroine[-6])
print(Heroine[1:])
print(Heroine[:5])
print(Heroine[:5:2])
print(Heroine[::-1])
print(Heroine[:5:3])
print(Heroine[:])
print(Heroine[7:1:1])
print(Heroine[7:1:-1])
print(Heroine[-1:-6:-1])
```

```python
print(Heroine[-1000:7:1])
```

```python
print(Heroine[0:2] + Heroine[5:])
```

```python
print(Heroine[::-1])
```

```python
# Reverse the string
l = [char for char in Heroine]
l.reverse()
print("".join(l))
```
