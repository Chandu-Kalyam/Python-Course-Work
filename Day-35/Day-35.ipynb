# Time Series based dataframes

```python
import pandas as pd
import numpy as np
```

```python
# date_range() pandas function is used for generating data related data.
date_rng = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = "YE")
date_rng
```

```python
date_rng = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = "YS")
date_rng
```

```python
date_rng = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = "ME")
date_rng
```

```python
date_rng = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = "D")
date_rng
```

```python
date_rng = pd.date_range(start = "2020-01-01", end = "2026-12-31", freq = "h")
date_rng
```

```python
# choose random 1000 date values
# random.choice() function is used to choose random values
# replace = False: don't choose the same value again
random_dates = np.random.choice(date_rng, size = 1000, replace = False)
random_dates.size
random_dates
```

```python
# Time series based dataframe: DataFrame with date/time column
# pd.to_datetime(): converting numpy date_time data into Pandas Data_time data
# np.arange(start, end, step): generating values from start to end - 1
# np.random.randint() : generating random integers : 20 to 199
df = pd.DataFrame(
    {
        "timestamp": random_dates,
        "order_id" : np.arange(1, 1001),
        "category" : np.random.choice(["Samosa", "EggPuff", "Coffee", "CongressMasala"], size = 1000),
        "price" : np.random.randint(20, 199, size = 1000),
        "quantity" : np.random.randint(1, 7, size = 1000)
    }
)
df["year"] = df.timestamp.dt.year
df["month"] = df.timestamp.dt.month
df["day"] = df.timestamp.dt.day

df["hours"] = df.timestamp.dt.hour
df["minute"] = df.timestamp.dt.minute
df["second"] = df.timestamp.dt.second
df
```

# chart building using pandas

```python
df.groupby("year")["quantity"].sum().plot(kind = "bar", rot = 45)
```

```python
df.groupby("year")["quantity"].sum().plot(kind = "pie", rot = 45)
```

```python
df.groupby("year")["quantity"].sum().plot(kind = "line", rot = 45)
```
