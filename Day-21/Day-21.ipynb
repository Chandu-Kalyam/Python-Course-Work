# Multilevel Inheritance 

```python
# GreatGF -> grandfather -> father -> son(child)
```

```python
class GreatGF:
    def __init__(self):
        print("I am Grand Grand father")
    
    def gold(self):
        print("1000 KG")
    
class Grandfather(GreatGF):
    def __init__(self):
        print("I am grand father")
        GreatGF.__init__(self)
        
    def land(self):
        print("500 Acres")

class Father(Grandfather):
    def __init__(self):
        print("I am father")
        Grandfather.__init__(self)
        
    def car(self):
        print("Ferrari")

class Son(Father):
    def __init__(self):
        print("I am son")
        Father.__init__(self)
        
    def bike(self):
            print("Hayabusa")

print("Son class")
s = Son()
print()
s.bike()
s.car()
s.land()
s.gold()
print()

print("Father class")
f = Father()
print()
f.car()
f.land()
f.gold()
print()

print("GrandFather class")
g = Grandfather()
print()
g.land()
g.gold()
print()

print("GreatGrandfather class can access:")
gf = GreatGF()
gf.gold()
```

# Hierarichal Inheritance

```python
# Many child one parent class
class Father:
    def __init__(self):
        print("I am calling my father")
        
    def gift(self):
        print("Gift is so much suprising i will reveal it soon.")


class Daughter(Father):
    def __init__(self):
        print("I am Daughter")
        Father.__init__(self)
    
    
class Son(Father):
    def __init__(self):
        print("I am Son")
        Father.__init__(self)
        
s = Son()
s.gift()

print()
d = Daughter()
d.gift()
```

```python
# Example-2: super() function
class Father:
    def __init__(self):
        print("I am calling my father")
        
    def gift(self):
        print("Gift is so much suprising i will reveal it soon.")


class Daughter(Father):
    def __init__(self):
        print("I am Daughter")
        super().__init__()
    
    
class Son(Father):
    def __init__(self):
        print("I am Son")
        super().__init__()
        
        
s = Son()
s.gift()

print()
d = Daughter()
d.gift()

# Method resolution order
print([i.__name__ for i in Son.__mro__])
print([i.__name__ for i in Daughter.__mro__])
```

# Hybrid Inheritance

```python
# example-1: structure of hybrid inheritance Diamond problem
class A:
    def __init__(self):
        print("I am A class")
        self.a_var = "A Class"
    

class B(A):
    def __init__(self):
        print("I am B class")
        self.b_var = "B Class"
        super().__init__()

class C(B):
    def __init__(self):
        print("I am C class")
        self.c_var = "C Class"
        super().__init__()

class D(B):
    def __init__(self):
        print("I am D class")
        self.d_var = "D Class"
        super().__init__()

class E(C, D):
    def __init__(self):
        print("I am E class")
        self.e_var = "E Class"
        super().__init__()
        

e = E()
print(e.e_var)


print()

print([i.__name__ for i in E.__mro__])
```
